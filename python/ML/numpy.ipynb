```python
# horizontal stacking

a = np.array([1,2,3])
b = np.array([4,5,6])
np.hstack((a,b))
```

```python
# vertical stacking

a = np.array([1,2,3])
b = np.array([4,5,6])

np.vstack((a,b))
```

```python
ar1 = np.array([1, 2, 3, 7, 8, 9])
ar2 = np.array([9, 8, 7, 6, 5, 4, 1])

intersection = np.intersect1d(ar1, ar2)
difference = np.setdiff1d(ar1, ar2)

print("Intersection: ", intersection)
print("Difference: ", difference)
```

```python
# difference between arrays and numpy arrays
# numpy arrays are faster than python arrays
# numpy arrays are more compact than python arrays
# numpy arrays are more convenient to use than python arrays
# numpy arrays allow for elementwise operations
# numpy arrays allow for broadcasting
# numpy arrays allow for vector and matrix operations

```
